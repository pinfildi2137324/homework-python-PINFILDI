```python
# ==========================================
# PROBLEM 1 - Alphabet Rangoli
# ==========================================

def print_rangoli(size):
    alphabet = "abcdefghijklmnopqrstuvwxyz"
    width= 4 * size - 3
    rows = []
    
    for i in range(size):
        right = alphabet[i:size]
        left = right[:0:-1]
        row = "-".join(left + right).center(width, "-")
        rows.append(row)
    for row in rows[::-1] + rows[1:]:
        print(row)
```

```python
# ==========================================
# PROBLEM 1 - Piling Up!
# ==========================================

from collections import deque

for _ in range(int(input())):
    n = int(input())
    d = deque(map(int, input().split()))
    
    top = float('inf')
    possible = True
    
    while d:
        if d[0] >= d[-1]:
            cube = d.popleft()
        else:
            cube = d.pop()
            
        if cube <= top:
            top = cube
        else:
            possible = False
            break
            
    print("Yes" if possible else "No")
```

```python
# ==========================================
# PROBLEM 1 - HTML Parser - Part 1
# ==========================================

from html.parser import HTMLParser

class MyHTMLParser(HTMLParser):
    def handle_starttag(self, tag, attrs):
        print(f"Start : {tag}")
        for name, value in attrs:
            print(f"-> {name} > {value}")

    def handle_endtag(self, tag):
        print(f"End   : {tag}")

    def handle_startendtag(self, tag, attrs):
        print(f"Empty : {tag}")
        for name, value in attrs:
            print(f"-> {name} > {value}")

n = int(input())
parser = MyHTMLParser()

for _ in range(n):
    parser.feed(input())
```

```python
# ==========================================
# PROBLEM 1 - Validating UID
# ==========================================

def is_valid_uid(uid):
    if len(uid) != 10 or len(set(uid)) != 10:
        return False
    if not uid.isalnum():
        return False
    if sum(1 for c in uid if c.isupper()) < 2:
        return False
    if sum(1 for c in uid if c.isdigit()) < 3:
        return False
    return True

t = int(input().strip())
for _ in range(t):
    uid = input().strip()
    if is_valid_uid(uid):
        print("Valid")
    else:
        print("Invalid")
```

```python
# ==========================================
# PROBLEM 1 - Validating Credit Card Numbers
# ==========================================

def is_valid_card(card):
    if '-' in card:
        parts = card.split('-')
        if len(parts) != 4 or any(len(p) != 4 for p in parts):
            return False
        card = ''.join(parts)
        
    if len(card) != 16 or not card.isdigit() or card[0] not in '456':
        return False
        
    if any(d * 4 in card for d in '0123456789'):
        return False
        
    return True

n = int(input().strip())
for _ in range(n):
    card = input().strip()
    if is_valid_card(card):
        print("Valid")
    else:
        print("Invalid")
```

```python
# ==========================================
# PROBLEM 1 - Matrix Script
# ==========================================

import math
import os
import random
import re
import sys




first_multiple_input = input().rstrip().split()

n = int(first_multiple_input[0])

m = int(first_multiple_input[1])

matrix = []

for _ in range(n):
    matrix_item = input()
    matrix.append(matrix_item)
decoded = ''.join(''.join(col) for col in zip(*matrix))

left = 0
while left < len(decoded) and not decoded[left].isalnum():
    left += 1

right = len(decoded) - 1
while right >= left and not decoded[right].isalnum():
    right -= 1

middle = ''.join(c * c.isalnum() or ' ' for c in decoded[left:right + 1])
print(decoded[:left] + ' '.join(middle.split()) + decoded[right + 1:])
```

```python
# ==========================================
# PROBLEM 2 - Decorators 2 - Viral Advertising
# ==========================================


import math
import os
import random
import re
import sys

def viralAdvertising(n):
    shared = 5
    cumulative = 0
    
    for _ in range(n):
        liked = shared // 2
        cumulative += liked
        shared = liked * 3
        
    return cumulative
```

```python
# ==========================================
# PROBLEM 2 - Decorators 2 - Recursive Digit Sum
# ==========================================

import math
import os
import random
import re
import sys

def superDigit(n, k):
    if len(n) == 1 and k == 1:
        return int(n)
    summing= sum(int(digit) for digit in n) * k
    return superDigit(str(summing), 1)

```

```python
# ==========================================
# PROBLEM 2 - Decorators 2 - Insertion Sort - Part 1
# ==========================================

import math
import os
import random
import re
import sys


def insertionSort1(n, arr):
    e = arr[n - 1]
    i = n - 2
    
    while i >= 0 and arr[i] > e:
        arr[i + 1] = arr[i]
        print(*arr)
        i -= 1
        
    arr[i + 1] = e
    print(*arr)
```

```python
# ==========================================
# PROBLEM 2 - Decorators 2 - Insertion Sort - Part 2
# ==========================================

import math
import os
import random
import re
import sys

def insertionSort2(n, arr):
    for i in range(1, n):
        e = arr[i]
        j = i - 1
        
        while j >= 0 and arr[j] > e:
            arr[j + 1] = arr[j]
            j -= 1
            
        arr[j + 1] = e
        print(*arr)
```
